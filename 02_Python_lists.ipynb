```python
print("PYTHON_LISTS")
print('=' * 80)


# Lists store multiple items in a single variable.
# List items are ordered & indexed, changeable, and allow duplicate items.
```

```python
# Construct a list
the_list = ['apple', 'orange', 'strawberry']
print(type(the_list))
print(the_list)

a_list = list(('apple', 34, True, 40, 'male')) # note the double round-brackets
print(type(a_list))
print(a_list)

print(len(the_list))
print(len(a_list))
```

```python
print('ACCESS_ITEMS')
print('=' * 80)

print(the_list[1])
print(a_list[-1]) # Negative indexing means start from the end.
print(a_list[2:5]) # gives items from second position excluding the item at fifth position.
print(a_list[:4]) # first four elements
print(a_list[2:]) # gives items from second position to the end.
print(the_list[-2:-1]) # starts from -2, excludes the -1 value.
```

```python
print('CHANGE, ADD, REMOVE LIST ITEMS')
print('=' *80)

the_list[1:2] = ['blackcurrent', 'guava']
print(the_list)
a_list[:3] = ['food']
print(a_list)
the_list[-3:] = ['strawberry']
print(the_list)

a_list.insert(2, 'watermelon') # to add an item at a specific position without replacing any.
print(a_list)

a_list.append('pineapple') # to add an item to the end of the list.
print(a_list)

a_list.extend(the_list) # to append elements from another list to the current list.
print(a_list)
```

```python
my_list = ['femy', 26, 'TECH', 'women', 'single', 'TECH' ]

my_list.remove('TECH') # removes the first occurrence.
print(my_list)

my_list.pop(3) # removes the specified index. If not specified, removes the last item.
print(my_list)

del my_list[2] # can also delete the list completely.
print(my_list)

my_list.clear() # empties the list.
print(my_list)
```

```python
print('LOOP THROUGH A LIST')
print('=' * 80)

print("for loop")
new_list = ["DL", "PHD", "skillset", "english", "confidence"]
for x in new_list:
  print(x)

print("while loop")
i = 0
while i < len(new_list):
  print(new_list[i])
  i = i + 1

```

```python
print('LIST COMPREHENSION')
print('=' * 80)

#Syntax = [expression for item in iterable if condition == True]
to_list = [print(x) for x in new_list if 'i' in x]
do_list = [print(x.upper()) for x in new_list]
up_list = [print(x) if x != 'skillset' else print('consistency') for x in new_list] #The expression can also contain conditions, not like a filter, but as a way to manipulate the outcome.
```

```python
print("SORT LISTS")
print('='*80)

t_list = ['january', 'February', 'March', 'June', 'october']


t_list.sort() #ascending
print(t_list)
t_list.sort(reverse = True) #descending
print(t_list)
t_list.reverse() #reversed order
print(t_list)
t_list.sort(key = str.lower) #to make sort() case-insensitive
print(t_list)
```

```python
print('COPY LISTS')
print('=' * 80)

o_list = t_list.copy()
print(o_list)
n_list = list(t_list)
print(n_list)
y_list = t_list[:]
print(y_list)
```

```python
print('JOIN LISTS')
print('='*80)

my_list = t_list + o_list
print(my_list)


femy_list = ['family', 'career', 'health', 'fitness']
for x in my_list:
  femy_list.append(x)
print(femy_list)


home_list = ['furniture', 'solar', 'appliances']
home_list.extend(y_list)
print(home_list)
```
